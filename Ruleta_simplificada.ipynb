```python
import random

def jugar_ruleta(opciones_ruleta, numeros_obtenidos):
    print("\n=========== RULETA SIMPLIFICADA ===========")

    # En este apartado se muestra el catálogo de opciones para apostar
    print("\n1. Rojo")
    print("2. Negro")
    print("3. Número")

    while True:
        try:
            opcion = int(input("Selecciona: "))
            if 1 <= opcion <= 3:
                break
            print("Selecciona una opción del 1 al 3.")
        except ValueError:
            print("Entrada inválida.")

    tipo_apuesta = opciones_ruleta[opcion - 1].upper()
    numero_elegido = None

    while True:
        try:
            apuesta = int(input("¿Cuántas fichas apuestas? "))
            if apuesta > 0:
                break
            print("La apuesta debe ser mayor que cero.")
        except ValueError:
            print("Ingresa una cantidad válida.")

    if tipo_apuesta == "NÚMERO":
        while True:
            try:
                numero_elegido = int(input("¿A qué número apuestas (1 - 36)? "))
                if 1 <= numero_elegido <= 36:
                    break
                print("El número debe estar entre 1 y 36.")
            except ValueError:
                print("Por favor, ingresa un número válido.")

    print("\nLa ruleta está girando...")
    numero_ruleta = random.randint(1, 36)
    numeros_obtenidos.add(numero_ruleta)
    print(f"Salió el número {numero_ruleta}.")

    # En este apartado se determina la regla simple de equivalencia de colores
    if numero_ruleta == 0:
        color_ruleta = "VERDE"
    elif numero_ruleta % 2 == 0:
        color_ruleta = "ROJO"
    else:
        color_ruleta = "NEGRO"

    print("\nResultado:")
    print(f"{numero_ruleta} - {color_ruleta}")

    gano = False
    premio = 0

    if tipo_apuesta == "ROJO" and color_ruleta == "ROJO":
        gano = True
        premio = apuesta
    elif tipo_apuesta == "NEGRO" and color_ruleta == "NEGRO":
        gano = True
        premio = apuesta
    elif tipo_apuesta == "NÚMERO" and numero_elegido == numero_ruleta:
        gano = True
        premio = apuesta * 5

    if gano:
        print(f"¡Ganaste {premio} fichas!")
    else:
        print(f"Perdiste {apuesta} fichas")

    return gano, premio, numero_ruleta




        
```
