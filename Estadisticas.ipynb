```python
class Jugador:
    """Clase para registrar los datos del jugador y su historial"""

    def __init__(self, nombre: str, fichas_iniciales: int = 100):
        self.nombre = nombre
        self.fichas = fichas_iniciales
        self.partidas_jugadas = 0
        self.victorias = 0
        self.derrotas = 0

        self.victorias_por_juego = {
            "Tragamonedas": 0,
            "Ruleta": 0,
            "Adivina el número": 0
        }

    def registrar_partidas(
        self,
        nombre_juego: str,
        gano: bool,
        fichas_ganadas_o_perdidas: int
    ):
        """Actualiza el historial general y por juego de cada ronda.
        - gano: True si el jugador ganó, False si perdió
        - fichas_ganadas_o_perdidas: entero positivo si ganó, negativo si perdió"""

        self.partidas_jugadas += 1
        self.fichas += fichas_ganadas_o_perdidas

        if gano:
            self.victorias += 1
            if nombre_juego in self.victorias_por_juego:
                self.victorias_por_juego[nombre_juego] += 1
        else:
            self.derrotas += 1


class EstadisticasCasino:
    """Clase para formatear y desplegar el reporte de estadísticas"""

    @staticmethod
    def mostrar_resumen(jugador: Jugador):
        """Tabla de estadísticas"""
        print("\n========= ESTADÍSTICAS =========")
        print(f"Jugador: {jugador.nombre}")
        print(f"Partidas jugadas: {jugador.partidas_jugadas}")
        print(f"Victorias: {jugador.victorias}")
        print(f"Derrotas: {jugador.derrotas}")
        print(f"Fichas actuales: {jugador.fichas}")

        # Este es el encabezado del desglose por juego
        print(f"{'Juego':<22} {'Victorias':<10}")
        print("-" * 32)

        formatear_linea = lambda juego, vics: f"{juego:<22} {vics:<10}"

        for juego, victorias in jugador.victorias_por_juego.items():
            print(formatear_linea(juego, victorias))

        print("=" * 32)
        
```
